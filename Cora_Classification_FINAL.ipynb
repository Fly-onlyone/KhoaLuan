```python

import warnings
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from rich.console import Console
from rich.table import Table
from sklearn.feature_selection import VarianceThreshold, SelectKBest, chi2
from sklearn.metrics import accuracy_score, f1_score, roc_curve, auc, classification_report, precision_score, recall_score
from sklearn.model_selection import train_test_split, RandomizedSearchCV, learning_curve
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import LabelEncoder, Normalizer, label_binarize
from sklearn.svm import LinearSVC, SVC
from sklearn.multiclass import OneVsRestClassifier
from imblearn.over_sampling import RandomOverSampler
from scipy.stats import uniform, randint

warnings.filterwarnings("ignore", category=RuntimeWarning)
warnings.filterwarnings("ignore", category=FutureWarning)
# Tắt convergence warnings để output sạch hơn
warnings.filterwarnings("ignore", message="Liblinear failed to converge")

# 1) Load data into DataFrame
from sklearn.preprocessing import StandardScaler

content = pd.read_csv("/kaggle/input/mycora/cora/cora.content", sep="\t", header=None)
ids = content.iloc[:, 0].values
X = content.iloc[:, 1:-1].values
labels = content.iloc[:, -1].values
le = LabelEncoder()
y = le.fit_transform(labels)
class_names = le.classes_

# ——— Cell A: Show first 10 lines of the raw data ———
import pandas as pd

content = pd.read_csv("/kaggle/input/mycora/cora/cora.content", sep="\t", header=None)
content.head(10)

# ——— Cell B: Demonstrate LabelEncoder mapping ———
from sklearn.preprocessing import LabelEncoder

labels = content.iloc[:, -1].values
le = LabelEncoder()
y = le.fit_transform(labels)

# In ra ánh xạ label → integer
for cls, code in zip(le.classes_, range(len(le.classes_))):
    print(f"{cls} → {code}")

# ===== BASELINE MODEL - Train ngay sau khi đọc dữ liệu =====
print("="*50)
print("TRAINING BASELINE MODEL (No parameter tuning, no data balancing)")
print("="*50)

# Split dữ liệu gốc (không cân bằng) cho baseline
X_train_baseline, X_test_baseline, y_train_baseline, y_test_baseline, id_train_baseline, id_test_baseline = train_test_split(
    X, y, ids, stratify=y, test_size=0.2, random_state=42
)

# Train baseline model với tham số mặc định
baseline_model = OneVsRestClassifier(LinearSVC(max_iter=5000, random_state=42))
baseline_model.fit(X_train_baseline, y_train_baseline)

# Evaluate baseline
y_pred_baseline = baseline_model.predict(X_test_baseline)
baseline_acc = accuracy_score(y_test_baseline, y_pred_baseline)
baseline_f1 = f1_score(y_test_baseline, y_pred_baseline, average='macro', zero_division=0)
baseline_prec = precision_score(y_test_baseline, y_pred_baseline, average='macro', zero_division=0)
baseline_rec = recall_score(y_test_baseline, y_pred_baseline, average='macro', zero_division=0)

print(f"Baseline Results:")
print(f"Accuracy: {baseline_acc:.4f}")
print(f"Precision (macro): {baseline_prec:.4f}")
print(f"Recall (macro): {baseline_rec:.4f}")
print(f"F1 Macro: {baseline_f1:.4f}")
print()

# 2) Plot distribution before upsampling
counts = pd.Series(y).value_counts().sort_index()
plt.figure(figsize=(8,4))
plt.bar(range(len(counts)), counts, edgecolor='black')
plt.xticks(range(len(counts)), class_names, rotation=45, ha='right')
plt.xlabel("Class Labels")
plt.ylabel("Number of Samples")
plt.title("Class Distribution Before Upsampling")
plt.tight_layout(); plt.show()

# 3) Combine into DataFrame for resampling
# Use all-string column names to avoid mixed types
df = pd.DataFrame(X)
df.columns = df.columns.astype(str)
df['id'] = ids
df['label'] = y

# 4) Upsample to balance classes
ros = RandomOverSampler(random_state=42)
df_res, y_res_series = ros.fit_resample(df, df['label'])
# convert resampled labels to numpy array to avoid pandas Series index issues
y_res = y_res_series.values

# 5) Plot distribution after upsampling
import matplotlib.pyplot as plt
import numpy as np

# Giả sử y là mảng nhãn gốc, y_res là mảng nhãn sau khi upsample
counts_orig = np.bincount(y)
counts_res  = np.bincount(y_res)
oversampled = counts_res - counts_orig

x = np.arange(len(counts_res))

plt.figure(figsize=(8,4))
# Vẽ phần gốc
plt.bar(x, counts_orig, edgecolor='black', label='Original')
# Vẽ phần tăng cường nằm chồng lên
plt.bar(x, oversampled, bottom=counts_orig, edgecolor='black',
        label='Oversampled', color='C1')  # C1 là màu mặc định thứ hai

plt.xticks(x, class_names, rotation=45, ha='right')
plt.xlabel("Class Labels")
plt.ylabel("Number of Samples")
plt.title("Class Distribution After Upsampling (Original vs. Oversampled)")
plt.legend()
plt.tight_layout()
plt.show()

# 6) Split features, ids, labels and do train/test for other models
X_resampled = df_res.drop(columns=['id','label']).values
ids_resampled = df_res['id'].values

X_train, X_test, y_train, y_test, id_train, id_test = train_test_split(
    X_resampled, y_res, ids_resampled, stratify=y_res, test_size=0.2, random_state=42
)

# Kiểm tra kích thước và đảm bảo không có cột 'id' trong X
print("Train shape:", X_train.shape)
print("Test  shape:", X_test.shape)

# Nếu DataFrame còn cột 'id', in ra tên cột để debug
if isinstance(X_train, pd.DataFrame):
    print("Train columns:", X_train.columns.tolist())

# 7) Define pipelines với chuẩn hóa và so sánh có/không có feature selection
from sklearn.preprocessing import StandardScaler

pipelines = {
    "SelectKBest + Normalizer": Pipeline([
        ("var_thresh", VarianceThreshold()),
        ("select", SelectKBest(chi2, k=300)),
        ("norm", Normalizer()),  # Thay StandardScaler bằng Normalizer
        ("clf", OneVsRestClassifier(LinearSVC(max_iter=20000, random_state=42)))
    ]),
    "Normalizer + No FS": Pipeline([
        ("var_thresh", VarianceThreshold()),
        ("norm", Normalizer()),  # Normalizer không có feature selection
        ("clf", OneVsRestClassifier(SVC(kernel='linear', probability=True, max_iter=20000, random_state=42)))
    ]),
    # So sánh: Model WITH feature selection
    "StandardScaler + FS": Pipeline([
        ("var_thresh", VarianceThreshold()),
        ("select", SelectKBest(chi2, k=500)),
        ("scaler", StandardScaler()),
        ("clf", OneVsRestClassifier(LinearSVC(max_iter=20000, tol=1e-4, C=0.1, dual=False, random_state=42)))
    ]),
    # So sánh: Model WITHOUT feature selection  
    "StandardScaler + No FS": Pipeline([
        ("var_thresh", VarianceThreshold()),
        ("scaler", StandardScaler()),
        ("clf", OneVsRestClassifier(LinearSVC(max_iter=20000, tol=1e-4, C=0.1, dual=False, random_state=42)))
    ])
}

dists = {
    "SelectKBest + Normalizer": {
        "select__k": randint(200, 600),
        "clf__estimator__C": uniform(0.01, 2.0),
        "clf__estimator__tol": uniform(1e-5, 1e-3)
    },
    "Normalizer + No FS": {
        "clf__estimator__C": uniform(0.01, 2.0),
        "clf__estimator__tol": uniform(1e-5, 1e-3)
    },
    "StandardScaler + FS": {
        "select__k": randint(300, 700),
        "clf__estimator__C": uniform(0.01, 1.0),
        "clf__estimator__tol": uniform(1e-5, 1e-3)
    },
    "StandardScaler + No FS": {
        "clf__estimator__C": uniform(0.01, 1.0),
        "clf__estimator__tol": uniform(1e-5, 1e-3)
    }
}

# 8) Hyperparameter tuning for other models
best_estimators = {}
# Thêm baseline model vào dict (đã train)
best_estimators["Baseline"] = baseline_model

for name, model in pipelines.items():
    model_to_tune = model if isinstance(model, Pipeline) else Pipeline([("clf", model)])
    rs = RandomizedSearchCV(
        model_to_tune, dists[name], n_iter=20, scoring='accuracy', cv=5,
        n_jobs=-1, verbose=1, random_state=42
    )
    rs.fit(X_train, y_train)
    best_estimators[name] = rs.best_estimator_
    print(f"{name} best params: {rs.best_params_}")

# 11) Classification reports and comparison table
console = Console()

# 1) Tìm model tốt nhất dựa trên F1-score và in classification report
best_model_name = None
best_f1_score = 0
best_model_results = {}

# Đánh giá baseline
baseline_f1_temp = f1_score(y_test_baseline, y_pred_baseline, average='macro', zero_division=0)
if baseline_f1_temp > best_f1_score:
    best_f1_score = baseline_f1_temp
    best_model_name = "Baseline"
    best_model_results = {
        'model': baseline_model,
        'y_test': y_test_baseline,
        'y_pred': y_pred_baseline,
        'test_set': 'baseline'
    }

# Đánh giá các model khác
for name, est in best_estimators.items():
    if name != "Baseline":
        y_pred = est.predict(X_test)
        f1_temp = f1_score(y_test, y_pred, average='macro', zero_division=0)
        if f1_temp > best_f1_score:
            best_f1_score = f1_temp
            best_model_name = name
            best_model_results = {
                'model': est,
                'y_test': y_test,
                'y_pred': y_pred,
                'test_set': 'balanced'
            }

# In classification report cho model tốt nhất
print(f"\n{'='*60}")
print(f"BEST MODEL: {best_model_name}")
print(f"F1-Score (macro): {best_f1_score:.4f}")
print(f"{'='*60}")
print(f"\nClassification Report for Best Model ({best_model_name}):")
print(classification_report(
    best_model_results['y_test'],
    best_model_results['y_pred'],
    target_names=class_names))

# 2) TẠO BẢNG SO SÁNH 
comparison = Table(show_header=True, header_style='bold magenta')
comparison.add_column('Pipeline')
comparison.add_column('Accuracy', justify='right')
comparison.add_column('Precision (macro)', justify='right')
comparison.add_column('Recall (macro)', justify='right')
comparison.add_column('F1 Macro', justify='right')

# Thêm baseline results
comparison.add_row(
    "Baseline",
    # "[red]No tuning/balancing[/red]", # <-- ĐÃ XÓA DỮ LIỆU CỘT TYPE
    f'{baseline_acc:.4f}',
    f'{baseline_prec:.4f}',
    f'{baseline_rec:.4f}',
    f'{baseline_f1:.4f}')

# Thêm results từ các model khác
for name, est in best_estimators.items():
    if name != "Baseline":  # Skip baseline
        y_pred = est.predict(X_test)
        acc   = accuracy_score(y_test, y_pred)
        prec  = precision_score(y_test, y_pred, average='macro', zero_division=0)
        rec   = recall_score(y_test, y_pred, average='macro', zero_division=0)
        f1    = f1_score(y_test, y_pred, average='macro', zero_division=0)
        comparison.add_row(
            name,
            # model_types.get(name, "[white]Other[/white]"), # <-- ĐÃ XÓA DỮ LIỆU CỘT TYPE
            f'{acc:.4f}',
            f'{prec:.4f}',
            f'{rec:.4f}',
            f'{f1:.4f}'
        )
console.print(comparison)

# Thêm phân tích so sánh Feature Selection với nhiều preprocessing methods
print("\n" + "="*60)
print("FEATURE SELECTION COMPARISON ANALYSIS")
print("="*60)

# Thu thập metrics cho các models có feature selection và không có
fs_comparison_data = []

# Models WITH Feature Selection
fs_models = [
    "SelectKBest + Normalizer",
    "StandardScaler + FS"
]

# Models WITHOUT Feature Selection  
no_fs_models = [
    "Normalizer + No FS",
    "StandardScaler + No FS"
]

# Tính metrics cho tất cả models
all_metrics = {}
# Baseline
all_metrics["Baseline"] = {
    'acc': baseline_acc,
    'prec': baseline_prec,
    'rec': baseline_rec,
    'f1': baseline_f1,
    'has_fs': False,
    'preprocessing': 'None'
}

# Các models khác
for name, est in best_estimators.items():
    if name != "Baseline":
        y_pred = est.predict(X_test)
        all_metrics[name] = {
            'acc': accuracy_score(y_test, y_pred),
            'prec': precision_score(y_test, y_pred, average='macro', zero_division=0),
            'rec': recall_score(y_test, y_pred, average='macro', zero_division=0),
            'f1': f1_score(y_test, y_pred, average='macro', zero_division=0),
            'has_fs': name in fs_models,
            'preprocessing': name.split(' + ')[1] if ' + ' in name else name.split(' ')[0]
        }

# Tạo bảng so sánh Feature Selection
fs_comparison_table = Table(show_header=True, header_style='bold blue')
fs_comparison_table.add_column('Model')
fs_comparison_table.add_column('Preprocessing', width=15)
fs_comparison_table.add_column('Feature Selection', width=12, justify='center')
fs_comparison_table.add_column('Accuracy', justify='right')
fs_comparison_table.add_column('Precision', justify='right')
fs_comparison_table.add_column('Recall', justify='right')
fs_comparison_table.add_column('F1-Score', justify='right')

# Thêm dữ liệu vào bảng, sắp xếp theo có/không có FS
for name, metrics in all_metrics.items():
    fs_status = "✅ Yes" if metrics['has_fs'] else "❌ No"
    preprocessing = metrics['preprocessing']
    
    # Màu sắc theo feature selection
    fs_color = "green" if metrics['has_fs'] else "red"
    
    fs_comparison_table.add_row(
        name,
        preprocessing,
        f"[{fs_color}]{fs_status}[/{fs_color}]",
        f"{metrics['acc']:.4f}",
        f"{metrics['prec']:.4f}",
        f"{metrics['rec']:.4f}",
        f"{metrics['f1']:.4f}"
    )

console.print("\nFeature Selection Impact Analysis:")
console.print(fs_comparison_table)

# Phân tích chi tiết theo preprocessing method
print(f"\n📊 DETAILED ANALYSIS:")

# So sánh StandardScaler models
std_with_fs = all_metrics.get("StandardScaler + FS", {})
std_no_fs = all_metrics.get("StandardScaler + No FS", {})

# So sánh Normalizer models
norm_with_fs = all_metrics.get("SelectKBest + Normalizer", {})
norm_no_fs = all_metrics.get("Normalizer + No FS", {})

# So sánh StandardScaler
if std_with_fs and std_no_fs:
    print(f"\n🔍 StandardScaler Models Comparison:")
    improvement_table = Table(show_header=True, header_style='bold cyan')
    improvement_table.add_column('Metric')
    improvement_table.add_column('With FS', justify='right')
    improvement_table.add_column('Without FS', justify='right')
    improvement_table.add_column('Improvement', justify='right')
    improvement_table.add_column('Status', justify='center')
    
    metrics_names = [('Accuracy', 'acc'), ('Precision', 'prec'), ('Recall', 'rec'), ('F1-Score', 'f1')]
    
    for metric_name, metric_key in metrics_names:
        with_val = std_with_fs[metric_key]
        without_val = std_no_fs[metric_key]
        improvement = with_val - without_val
        status = "🟢 Better" if improvement > 0 else "🔴 Worse" if improvement < 0 else "🟡 Same"
        
        improvement_table.add_row(
            metric_name,
            f"{with_val:.4f}",
            f"{without_val:.4f}",
            f"{improvement:+.4f}",
            status
        )
    
    console.print(improvement_table)

# So sánh Normalizer
if norm_with_fs and norm_no_fs:
    print(f"\n🔍 Normalizer Models Comparison:")
    norm_improvement_table = Table(show_header=True, header_style='bold magenta')
    norm_improvement_table.add_column('Metric')
    norm_improvement_table.add_column('With FS', justify='right')
    norm_improvement_table.add_column('Without FS', justify='right')
    norm_improvement_table.add_column('Improvement', justify='right')
    norm_improvement_table.add_column('Status', justify='center')
    
    for metric_name, metric_key in metrics_names:
        with_val = norm_with_fs[metric_key]
        without_val = norm_no_fs[metric_key]
        improvement = with_val - without_val
        status = "🟢 Better" if improvement > 0 else "🔴 Worse" if improvement < 0 else "🟡 Same"
        
        norm_improvement_table.add_row(
            metric_name,
            f"{with_val:.4f}",
            f"{without_val:.4f}",
            f"{improvement:+.4f}",
            status
        )
    
    console.print(norm_improvement_table)
    
    # Feature selection info
    if std_with_fs:
        with_fs_model = best_estimators["StandardScaler + FS"]
        selected_features = with_fs_model.named_steps['select'].get_support().sum()
        total_features = X_train.shape[1]
        print(f"\n📋 Features selected: {selected_features}/{total_features} ({selected_features/total_features*100:.1f}%)")

# So sánh các preprocessing methods với feature selection
print(f"\n🔄 Preprocessing Methods Comparison (with Feature Selection):")
preprocessing_table = Table(show_header=True, header_style='bold magenta')
preprocessing_table.add_column('Preprocessing Method')
preprocessing_table.add_column('Model Name', width=30)
preprocessing_table.add_column('F1-Score', justify='right')
preprocessing_table.add_column('Accuracy', justify='right')

fs_only_models = {name: metrics for name, metrics in all_metrics.items() 
                  if metrics['has_fs'] and name != "Baseline"}

# Sắp xếp theo F1-score
sorted_fs_models = sorted(fs_only_models.items(), key=lambda x: x[1]['f1'], reverse=True)

for name, metrics in sorted_fs_models:
    preprocessing_table.add_row(
        metrics['preprocessing'],
        name,
        f"{metrics['f1']:.4f}",
        f"{metrics['acc']:.4f}"
    )

console.print(preprocessing_table)

print("="*60)

# 12) Illustrate 10 test samples: paper ID, true, pred, features
# Sử dụng baseline test set
idx = np.random.choice(len(X_test_baseline), 10, replace=False)
feature_table = Table(show_header=True, header_style='bold cyan')
feature_table.add_column('Paper ID', width=12)
feature_table.add_column('True Label', width=22)
feature_table.add_column('Predicted', width=22)
feature_table.add_column('Vector', overflow='fold')
for i in idx:
    pid = id_test_baseline[i]
    feat_str = np.array2string(X_test_baseline[i], precision=0, separator=',', max_line_width=1000)
    pred_label = baseline_model.predict([X_test_baseline[i]])[0]
    feature_table.add_row(
        str(pid), class_names[y_test_baseline[i]], class_names[pred_label], feat_str
    )
console.print(feature_table)
```
